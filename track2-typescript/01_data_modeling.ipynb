# 문항 1: 전자상거래 회원·상품 데이터 모델링 및 타입 안전 집계 유틸리티 구현

## 1-1. 회원 모델 정의

```typescript
// 권한을 정의하는 문자열 리터럴 유니언 타입
type UserRole = 'admin' | 'customer' | 'seller';

// 회원 인터페이스
interface User {
  id: number;
  username: string;
  email: string;
  fullName: string;
  role: UserRole;
  createdAt: Date;
  phone?: string;  // optional
}

console.log('User interface defined');
```

## 1-2. 상품 모델 정의

```typescript
// 카테고리를 제한하는 타입 별칭
type Category = 'electronics' | 'clothing' | 'food' | 'furniture' | 'books';

// 상품 인터페이스
interface Product {
  id: number;
  title: string;
  description: string;
  price: number;
  category: Category;
  stock: number;
  isActive: boolean;
}

console.log('Product interface defined');
```

## 1-3. 더미 데이터 생성 및 검증

```typescript
// 더미 회원 배열
const users: User[] = [
  {
    id: 1,
    username: 'admin_user',
    email: 'admin@example.com',
    fullName: '관리자',
    role: 'admin',
    createdAt: new Date('2024-01-01'),
    phone: '010-1234-5678',
  },
  {
    id: 2,
    username: 'kim_customer',
    email: 'kim@example.com',
    fullName: '김고객',
    role: 'customer',
    createdAt: new Date('2024-03-15'),
  },
  {
    id: 3,
    username: 'lee_seller',
    email: 'lee@example.com',
    fullName: '이판매자',
    role: 'seller',
    createdAt: new Date('2024-05-20'),
    phone: '010-9876-5432',
  },
];

// 더미 상품 배열
const products: Product[] = [
  {
    id: 1,
    title: '스마트폰 케이스',
    description: '고품질 가죽 케이스',
    price: 25000,
    category: 'electronics',
    stock: 100,
    isActive: true,
  },
  {
    id: 2,
    title: '겨울 패딩 점퍼',
    description: '따뜻한 겨울 패딩',
    price: 189000,
    category: 'clothing',
    stock: 50,
    isActive: true,
  },
  {
    id: 3,
    title: '유기농 쌀 5kg',
    description: '국내산 유기농 쌀',
    price: 32000,
    category: 'food',
    stock: 200,
    isActive: false,
  },
];

console.log('Users:', users);
console.log('Products:', products);
```

## 1-4. 타입 안전 집계 유틸리티

```typescript
// 역할별 사용자 필터링
const getUsersByRole = (users: User[], role: UserRole): User[] =>
  users.filter(u => u.role === role);

// 카테고리별 상품 필터링
const getProductsByCategory = (products: Product[], category: Category): Product[] =>
  products.filter(p => p.category === category);

// 활성 상품의 평균 가격
const getAveragePrice = (products: Product[]): number => {
  const active = products.filter(p => p.isActive);
  if (active.length === 0) return 0;
  return active.reduce((sum, p) => sum + p.price, 0) / active.length;
};

// 테스트
console.log('Admins:', getUsersByRole(users, 'admin'));
console.log('Electronics:', getProductsByCategory(products, 'electronics'));
console.log('Average price:', getAveragePrice(products));
```
